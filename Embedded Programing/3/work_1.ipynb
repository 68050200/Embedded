**ตอนที่ 3: Two's complement**

```python
#68050200

#7.หา two's complement 8 ปิตของ -25 (แสดงขั้นตอน invert + 1)
print("7.หา two's complement 8 ปิตของ -25 (แสดงขั้นตอน invert + 1)")
neg = (-25) & 0xFF
print("bit pattern ของ -25 :", format(neg, '08b'))
print ("25 = 16 + 8 + 1 \n" \
"   = 0001 1001\n" \
"     invert \n" \
"   = 1110 0110   (one's complement)\n" \
"   = 1110 0111   (two's complement) invert + 1")

#8.int8_t มีบิต 10000001 มีค่าเท่าใตในเลขฐานสิบ
print("\n8.int8_t มีบิต 10000001 มีค่าเท่าใตในเลขฐานสิบ")
def to_int8(bits_value):

    v = bits_value & 0xFF
    return v - 256 if v > 127 else v

print("10000001 =", to_int8(0b10000001))   

#9.จงหาผลลบ 5 - 8 โดยใช้การบวก two's complement ขนาด 8 บิต แล้วแปลงผลกลับเป็นเลขฐานสิบ
print("\n9.จงหาผลลบ 5 - 8 โดยใช้การบวก two's complement ขนาด 8 บิต แล้วแปลงผลกลับเป็นเลขฐานสิบ")

#แปลงเป็น two's complement
A = (5) & 0xFF
B = (-8) & 0xFF
print("A =", format(A, '08b'))
print("B =", format(B, '08b'))
print("00000101 =", to_int8(0b00000101))
print("11111000 =", to_int8(0b11111000))

print("A + B  =", format(A+B, '08b'))
print("A + B  ", to_int8(A+B))
```
