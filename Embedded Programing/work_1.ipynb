**ตอนที่ 1:การแปลงเลขฐาน**

```python
#1.แปลงฐานสองเป็นฐานสิบ: 11010110 
print("1.แปลงฐานสองเป็นฐานสิบ: 11010110 ")
print(int("11010110", 2))

#2.แปลงฐานสิบเป็นฐานสอง 8 บิต: 175
print("2.แปลงฐานสิบเป็นฐานสอง 8 บิต: 175 ")
print(bin(175)[2:])

#3.แปลงฐานสิบหกเป็นฐานสิบ: 0x3c และ 0xFF
print("3.แปลงฐานสิบหกเป็นฐานสิบ: 0x3c และ 0xFF")
print("0x3c = ", 0x3c)
print("0xFF = ", 0xFF)

#4.แปลงฐานสิบเป็นฐานสิบหก: 237 และ 4096
print("4.แปลงฐานสิบเป็นฐานสิบหก: 237 และ 4096")
print("237 =", hex(237))
print("4096 =", hex(4096))

```

**ตอนที่ 2: ช่วงข้อมูลและ overflow**

```python
#5.uint16_t (16 บิต ไม่มีเครื่องหมาย) เก็บค่าได้ช่วงใด และค่ามากสุดเท่าใด
print("5.uint16_t (16 บิต ไม่มีเครื่องหมาย) เก็บค่าได้ช่วงใด และค่ามากสุดเท่าใด")
u = 65530
for _ in range(8):
    print(u)
    u = (u + 1) & 0xFFFF

print("เก็บได้ช่วง0 ถึง 65535 \nค่ามากสุดคือ 65535\n")

#6.uint8_t a = 200; a = a + 100; ค่า a สุดท้ายเป็นเท่าใด อธิบายเหตุผล


print("6.uint8_t a = 200; a = a + 100; ค่า a สุดท้ายเป็นเท่าใด อธิบายเหตุผล")
a = 200
a = (a + 100) & 0xFF

print(f"ค่า a สุดท้ายคือ {a} เพราะเกินขอบเขตสูงสุดที่ 8 บิตเก็บได้ (255)")
```
